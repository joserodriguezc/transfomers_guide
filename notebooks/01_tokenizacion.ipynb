Comparamos los tokenizadores de BERT, GPT-2 y Qwen sobre el mismo texto en español e inglés y medimos su **fertilidad** (tokens por palabra).

```python
from transformers import AutoTokenizer

tokenizadores = {
    "BERT": AutoTokenizer.from_pretrained("bert-base-uncased"),
    "GPT-2": AutoTokenizer.from_pretrained("gpt2"),
    "Qwen2.5": AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B"),
}
```

```python
textos = {
    "inglés": "The Transformer has revolutionized NLP.",
    "español": "El transformer ha revolucionado el procesamiento del lenguaje natural.",
}
for nombre, tok in tokenizadores.items():
    for idioma, t in textos.items():
        piezas = tok.tokenize(t)
        print(f"{nombre:8s} {idioma:8s} {len(piezas):3d} tokens  fertilidad {len(piezas) / len(t.split()):.2f}  {piezas}")
```

```python
# IDs con tokens especiales: [CLS] = 101 y [SEP] = 102 en BERT
ids = tokenizadores["BERT"](textos["inglés"])["input_ids"]
print(ids)
print(tokenizadores["BERT"].convert_ids_to_tokens(ids))
```

```python
# Las primeras fusiones BPE aprendidas por GPT-2
tok = tokenizadores["GPT-2"]
fusiones = tok.backend_tokenizer.model.__class__.__name__
print("modelo del tokenizador:", fusiones, "| tamaño del vocabulario:", tok.vocab_size)
print([tok.decode([i]) for i in range(256, 276)])
```
