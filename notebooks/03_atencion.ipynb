# Matrices de atención de BERT

Notebook opcional del **módulo 3** del curso (en la app: `uv run dev` y la ruta `#/modulo/atencion`). Usa modelos reales; el módulo interactivo explica la teoría con NumPy.

> **Requisitos:** desde la raíz del repositorio, `uv sync --group notebooks` y abre este notebook con `uv run --group notebooks jupyter lab`. La primera ejecución descarga los modelos de Hugging Face.

```python
import matplotlib.pyplot as plt
import torch
from transformers import AutoModel, AutoTokenizer

tok = AutoTokenizer.from_pretrained("bert-base-uncased")
modelo = AutoModel.from_pretrained("bert-base-uncased", output_attentions=True, attn_implementation="eager")
```

```python
def atencion_de(frase, palabra, capa=8):
    x = tok(frase, return_tensors="pt")
    with torch.no_grad():
        att = modelo(**x).attentions[capa][0].mean(0)   # media de las cabezas
    tokens = tok.convert_ids_to_tokens(x["input_ids"][0])
    i = tokens.index(palabra)
    return tokens, att[i].numpy()

for frase in ["The bank of the river was full of ducks.", "The bank approved the loan."]:
    tokens, pesos = atencion_de(frase, "bank")
    plt.figure(figsize=(8, 2))
    plt.bar(tokens, pesos)
    plt.title(frase)
    plt.xticks(rotation=45)
    plt.show()
```
