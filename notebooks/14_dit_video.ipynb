Notebook opcional del **módulo 14** del curso (en la app: `uv run dev` y la ruta `#/modulo/dit-video`). Usa modelos reales; el módulo interactivo explica la teoría con NumPy.

> **Requisitos:** desde la raíz del repositorio, `uv sync --group notebooks` y abre este notebook con `uv run --group notebooks jupyter lab`. La primera ejecución descarga los modelos de Hugging Face.

> ⚠️ **Requiere GPU** con CUDA para ejecutarse en un tiempo razonable.

```python
import torch
from diffusers import DiTPipeline

pipe = DiTPipeline.from_pretrained("facebook/DiT-XL-2-256", torch_dtype=torch.float16).to("cuda")
ids = pipe.get_label_ids(["golden retriever", "volcano"])
imagenes = pipe(class_labels=ids, num_inference_steps=25, guidance_scale=4.0).images
imagenes[0]
```

```python
from diffusers import PixArtSigmaPipeline

pixart = PixArtSigmaPipeline.from_pretrained("PixArt-alpha/PixArt-Sigma-XL-2-512-MS", torch_dtype=torch.float16).to("cuda")
pixart("a cozy reading nook with books about transformers, warm light", num_inference_steps=20).images[0]
```

Vídeo con LTX-Video (unos 2 segundos a 24 fps). Requiere una GPU con bastante memoria.

```python
from diffusers import LTXPipeline
from diffusers.utils import export_to_video

ltx = LTXPipeline.from_pretrained("Lightricks/LTX-Video", torch_dtype=torch.bfloat16).to("cuda")
video = ltx(prompt="a paper boat floating on a calm river at sunset", width=704, height=480, num_frames=49,
            num_inference_steps=40, guidance_scale=3.0).frames[0]
export_to_video(video, "barco.mp4", fps=24)
```
