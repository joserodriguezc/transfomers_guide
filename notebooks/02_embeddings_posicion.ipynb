```python
import torch
import matplotlib.pyplot as plt
from transformers import AutoModel, AutoTokenizer
```

Los embeddings contextuales de BERT para una frase.

```python
tok = AutoTokenizer.from_pretrained("bert-base-uncased")
modelo = AutoModel.from_pretrained("bert-base-uncased")
entrada = tok("The Transformer has revolutionized NLP.", return_tensors="pt")
print(entrada["input_ids"])
with torch.no_grad():
    salida = modelo(**entrada)
print(salida.last_hidden_state.shape)
```

Las posiciones **aprendidas** de GPT-2: similitud coseno entre los vectores de las primeras 128 posiciones.

```python
gpt2 = AutoModel.from_pretrained("gpt2")
P = gpt2.wpe.weight[:128].detach()
P = P / P.norm(dim=1, keepdim=True)
plt.imshow((P @ P.T).numpy(), cmap="viridis")
plt.title("Coseno entre posiciones aprendidas (GPT-2)")
plt.colorbar()
plt.show()
```
