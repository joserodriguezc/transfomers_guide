# Encoder, decoder y encoder-decoder en acción

Notebook opcional del **módulo 6** del curso (en la app: `uv run dev` y la ruta `#/modulo/encoder-decoder`). Usa modelos reales; el módulo interactivo explica la teoría con NumPy.

> **Requisitos:** desde la raíz del repositorio, `uv sync --group notebooks` y abre este notebook con `uv run --group notebooks jupyter lab`. La primera ejecución descarga los modelos de Hugging Face.

```python
from transformers import pipeline
```

**Encoder (BERT):** rellenar un hueco mirando a ambos lados.

```python
relleno = pipeline("fill-mask", model="bert-base-uncased")
for r in relleno("The transformer [MASK] all tokens in parallel.")[:3]:
    print(r["token_str"], round(r["score"], 3))
```

**Decoder (GPT-2):** continuar un texto. Compara la decodificación voraz con el muestreo con temperatura y top-p.

```python
gen = pipeline("text-generation", model="gpt2")
prompt = "Transformers are useful because"
print(gen(prompt, max_new_tokens=30, do_sample=False)[0]["generated_text"])
for _ in range(2):
    print(gen(prompt, max_new_tokens=30, do_sample=True, temperature=0.8, top_p=0.9)[0]["generated_text"])
```

**Encoder-decoder:** traducción inglés → español.

```python
traductor = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")
print(traductor("The decoder attends to the encoder through cross-attention.")[0]["translation_text"])
```
