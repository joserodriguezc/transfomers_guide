```python
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller

co2 = sm.datasets.co2.load_pandas().data["co2"].resample("MS").mean().interpolate()
co2.plot(title="CO₂ en Mauna Loa")
plt.show()
```

```python
descomp = sm.tsa.seasonal_decompose(co2, model="additive", period=12)
descomp.plot()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
plot_acf(co2, lags=48, ax=ax[0])
plot_pacf(co2.diff().dropna(), lags=48, ax=ax[1])
plt.show()
```

```python
for nombre, serie in [("original", co2), ("diferenciada", co2.diff().dropna()), ("dif. estacional", co2.diff().diff(12).dropna())]:
    p = adfuller(serie)[1]
    print(f"{nombre:16s} p-valor de Dickey-Fuller = {p:.4f}", "(estacionaria)" if p < 0.05 else "")
```
